````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris 

iris = load_iris() 

x = iris.data
y= iris.target

print('shape of x : ', x.shape)
print('shape of y : ', y.shape)





````

````python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

#30 tasho --> test , 120 --> training datam




````

````python
from sklearn.neural_network import MLPClassifier 

#MLP --> Multi Layer Perceptron --> chand laye haye perceptron 
#MLP CLASSFIER, MLP REGRESSOR 

#vorodi  , khoroji , hidden layer


#[4 --> --> --> --> 3]

#4 ta vorodi (x hat)

#3 -> 3 ta gaol gol0 ,1 , 2

#yek layer ba 10 neuron

#activation --> sigmoid, relu , tanh -->hyperparamter

model = MLPClassifier(hidden_layer_sizes= (10,) ,activation='relu',max_iter=500,random_state=42 )

#aval random entkheab, man seedesho set mikonm , 


model.fit(x_train,y_train)

y_pred = model.predict(x_test)



from sklearn.metrics import accuracy_score

test_accuracy = accuracy_score(y_test,y_pred)

print(f"Test Accuracy : {test_accuracy}")



````
