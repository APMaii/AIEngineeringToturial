```python
import torch
import torch.nn as nn

class RNNClassifier(nn.Module):
    def __init__(self,vocab_size , embedding_dim, hidden_size):
        super().__init__()


        self.embedding = nn.Embedding(vocab_size,embedding_size)  #yek hoirod ya harchizi --> vector ha 

        self.rnn = nn.RNN(input_size=embedding_size , hidden_size = hidden_size, batch_first = True )

        self.fc= nn.Linear(hidden_size, 1)

        self.sigmoid = nn.sigmoid()


    def forward(self,x):

        #token (words)--> Vectors (number)
        #text --> numbers
        x = self.embedding(x)

        output , hidden = self.rnn(x)

        h_final = hidden[-1] #hidden akahri 

        logits = self.fc(h_final) #--> 1 -> harchi mide --> roo inam

        return self.sigmoid(logits) #

#yek jomle --> yek adadi 0 -1 --> 
#sentiment analysis 

#analyze ehsasate yek jomle ro 

#0 ta 1 --> 0 negative, 1 positive 
```
