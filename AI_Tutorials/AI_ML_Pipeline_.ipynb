```python


#step 0 ----> data ro import mikonid , clean , x , y tabdil mikonid
#pd.read_csv, read_excel , download , sklearn.datasset , .....


#pd.dropna, fillna, duplciate, if , else ,...


#x=np.array
#y=np


#step 1 --> train test

from sklearn.model_selection import train_test_split
x_train , x_test , y_train, y_test = train_test_split(x,y,test_size=0,2 , shuffle=True, random_state = 42)


# step 2 --> sakhele pipeline (model , processor)

from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers = [

        ('cat',OneHotEncoder(),['City'] ),
        ('num',StandardScaler(),['Age','Salary'])

    ]
)

#from sklearn.preprocess import PolynomialFeatures
#poly = PolynomialFeatures(degree=2)


from sklearn.svm import SVR

model = SVR()



from sklearn.pipeline import Pipeline

pipe = Pipeline( [ ('preprocesor',preprocessor ), ('model',model)])


#pipe miad hameye inaro kenare ham gahrar mide va tamame tavabeshon ham dare , fit fit ras , model --> pipe

#moel.fit , gabesh scaler.fit , ...

#pipe -->pipe.fit -->hamashon posht sare ham ba nazm khodesh 

#gridsearch(estimator -->model ,pipe)

# step 3 --> gridsearch

#ghadim --> C = [0.00001,0.0001,0.001,0.01,0.1,1,10]


#model nadari fgth , pipe (too delesh model)


hyperparametrs = { 'model__C': [0.00001,0.0001,0.001,0.01,0.1,1,10],
               'preprocesor'=[preprocessor ,StandardScaler(),MinMaxScaler()]}

from sklearn.model_selection import GridSearchCV

grid = GridSearchCV(
    estimator = pipe,
    param_grid = hyperparametrs ,
    scoring = 'neg_mean_absolute_error',
    cv= 5 
    #n_jobs = 5 , -1 parallel 
    #refit = True --> bad az inke hame kar krdi , roye kole data fitesh grid.predict

    #return_train_score=True  , test_score misanje ama ino true inam dar nazar migire
)


grid.fit(x_train,y_train)


grid.best_params_
#behtrain paramter

grid.best_score_


#Last step

from sklearn.metrics import mean_absolute_error

y_pred = grid.predict(x_test)

mae_test_score = mean_absolute_error(y_train, y_pred)


#akahrin resultet


```
