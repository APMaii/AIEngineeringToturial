```python



#step 0.0 import data

#step 0.1 Cleaning the data (empty cell, duplicated , type , logical ) drop , fill ,...

# step 0.2 Tbadil b x , y (numpy array)


# step 1 Train Test split
from sklearn.model_selection import train_test_split

x_train,x_test, y_train, y_test = train_test_split(x,y,test_size=0.2 , shuffle=True , random_state= 42)


# satep 2 model selction
#---------regressor-----------
from sklearn.linear_model import SGDRegressor
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensembles import RandomForestRegressor
from sklearn.svm import SVR


#configuration
model = SVR(C=10 , gamma= 0.001 , kernel = 'poly' , degree= 3)



#--------classifier --------------
from sklearn.linear_model import SGDClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensembles import RandomForestClassifier
from sklearn.svm import SVC

model = RandomForestClassifier(n_estimators = 200 , max_depth= 4)


# step 3 - trianing 
#man taghsim kardam datamo b yek train dataset , test dataset 
#modelamo configure krdm va train mikonm fght rooye train dataset 
model.fit(x_train,y_train)


# step 4 - prediction
y_pred_train = model.predict(x_train)  #x hae k dide ro bede behem


y_pred_test = model.predict(x_test)   #x test hamo midam b model , migam pishbini koij



# step 5 - Evaluation 

#---------regressor-----------
from sklearn.metrics import mean_absolute_error #mae
from sklearn.metrics import mean_absolute_percentage_error #mape
from sklearn.metrics import mean_squared_error #mse
from sklearn.metrics import r2_score # vali nesbt b sherkat, maghale 
from sklearn.metrics import explained_variance_score 





#-----classifier------------------
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import  recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import roc_auc_score


#inha mige beman true value , predicted value bede , man b raveshe khodam hesab mikonm

train_score_mae = mean_absolute_error(y_train,y_pred_pred) #y_train , x_train-->model-->y_pred_train  

test_score_mae = mean_absolute_error(y_test, y_pred_test)


#train score --> neshon mide k cheghadr khoob modelet yad grfte
#test score --> neshon mide k cheghadr modelet khob mitone pishbini kone


#ag Test scoret bala bood , ok tamam shod , ama age paein bood -> trade off

# negah koni b Train score 


#Train score kh paein bashe --> underfitting , dalile inke test score paeine, train score , yani asan model yad nagerefte  --> model ro model complex , too khode model biay az parametr hae estefade koni k bre samte complex shodan , strict shodan 


# Train scoret kh bala bood --> Overfitting , dalile inke test score paein, modele man k khob yad grfte pas chra nmitone predict (pishbini) , midek overfit shode hata error haram yad grfte, fght az train dataset yad grfte va bias shode , general nist . berid az complexity fasele begirid va berid samte geenralization. az penalty (L1,L2) , C in SVM , smOOTH tar yad bgire , az overfitting fasele begiri  


# shoma niaz dari k chan goone configuratione yek model ro hey train koni va test koni 








## GrdiSearch

yadavari --> hyperparameter, parametr haee hast zamani k model selection tarif mikoni va mdoel oonahro yad nemigire balke in paramter ha hatsan k rooye yadgirie model tasir daran . 


gridsearch yani yek model ro beri dar tarkibe hyperparametrash dar range haye mokhtaleft hey train mikone, hey test mikone va test score


SVR (C=0.0001) --> svr.fit(train) test_score --> 0.87 <br>
SVR (C=0.001) --> svr.fit(train) test_score --> 0.83 <br>
SVR(C=0.01) --> svr.fit(train) test_score --> 0.81 <br>
SVR(C=0.1) --> svr.fit(train) test_score --> 0.94 <br>
SVR(C=1) ---> svr.fit(train) test_score --> 0.92 <br>
SVR(10) --> svr.fit(train) test_score --> 0.91 <br>



chanta parameter , 7  (mohem tarin parameter)

SVR (kernel= 'Linear' , C=0.0001) --> svr.fit(train) test_score --> 0.87 <br>
SVR (kernel= 'Linear' ,C=0.001) --> svr.fit(train) test_score --> 0.83 <br>
SVR(kernel= 'Linear' ,C=0.01) --> svr.fit(train) test_score --> 0.81 <br>
SVR(kernel= 'Linear' ,C=0.1) --> svr.fit(train) test_score --> 0.83 <br>
SVR(kernel= 'Linear' ,C=1) ---> svr.fit(train) test_score --> 0.82 <br>
SVR(kernel= 'Linear' ,10) --> svr.fit(train) test_score --> 0.84 <br>
SVR (kernel= 'Poly' , C=0.0001) --> svr.fit(train) test_score --> 0.91 <br>
SVR (kernel= 'poly' ,C=0.001) --> svr.fit(train) test_score --> 0.94 <br>
SVR(kernel= 'poly' ,C=0.01) --> svr.fit(train) test_score --> 0.81 <br>
SVR(kernel= 'poly' ,C=0.1) --> svr.fit(train) test_score --> 0.83 <br>
SVR(kernel= 'poly' ,C=1) ---> svr.fit(train) test_score --> 0.95 <br>
SVR(kernel= 'poly' ,10) --> svr.fit(train) test_score --> 0.96 <br>





```python


x_train, x_test , y_train,y_test = train_test_split(x,y,test_size=0.25,shuffle=True, random_state=42 )


from skelarn.svm import SVR
from sklearn.metrics import mean_absolute_error

gridsearch_results= []

for c_value in [0.0001,0.001,0.01,0.1,1,10]:
    model = SVR(kernel='linear', C= c_value)
    model.fit(x_train,y_train)

    y_pred_train= model.predict(x_train)
    y_pred_test = model.predict(x_test)


    train_score= mean_absolute_error(y_train,y_pred_train)
    test_score = mean_absolute_error(y_test , y_pred_test)


    gridsearch = {'C':c_value , 'train_score' :train_score , 'test_score' : test_score}

    gridsearch_results.append(gridsearch)




minimum_mae = 10000000000
best_c = None

for results in gridsearch_results:

    test_score = results['test_score']

    if test_score < minimum_mae:
        minimum_mae = test_score
        best_c = results['C']


print(f'the best model in the range of [0.0001 , 10 ] best C is {best_c} with MAE of {minimum_mae}')
```
