zamani k datamon ro besoorate batch batch 32 ta 32 ta mifrestim, input layer be tedade feature (vizhegi) ha neuron dare, adad miran dakhel , zarb dar W weight (vazn) mishan ta b layere 1 hidden layer beresan + bias mishavadnd. dakhele neuron haye layere 1 , ina sum() + bias mishaavnad va mriavand zire yek activation function k in non linear mikonad , va oinakr edame peyda mikonad dakhele hidden layers ta berese be output layer.

Back propagation

miad nesbat b vazne laye haye ghabli error ro montasher mikone ke har neuron chegahdr dar in khata sahm dashte , va sepas miad moshtaghe loss bar weight haro hesab mikone. ba estefade az ghanone zanjire chain rules va sepas ma vaghty moshtaghe loss be weigth ro dahste bashim ya loss b bias , mitonim dar on iteration va epoch update konim weight va bais hamon ro

pas formula ine ke $$W_{new} = W_{old} + \frac{ \partial{Loss}}{\partial{W}}$$
 $$B_{new} = B_{old} + \frac{ \partial{Loss}}{\partial{B}}$$

 pas az chanidn iteration va epoch , baes mishavad neural networke ma weigth ha va biasi ro peyda kone k kamtarin loss ro dahste bashe ya behtare begim kamtarin faseley beyen khoroji va vagheiat --> **optimzier** ya mitonim begim k dige in network (Neural netwok) ba in weight o bais ha vaghty vorodi toosh bezari, khoroji e k mide  vageiat kheyli nazdike --> pishbini anajm dahi
