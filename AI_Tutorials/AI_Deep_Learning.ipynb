````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2,shuffle=True, random_state=42,stratify=y)

#20% test s kon , 80% train kon. random shuffler kon
#havaset bashe beham rikhte hast datam , train test , hame badkhimaro narizi too test 

#toye data unbalance 50% badkhim 50% khosh khim --> stratify=y


#scale --> Pipeline 

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.neural_network import MLPClassifier

model = Pipeline([
    ('scaler',StandardScaler()),
    ('mlp',
    MLPClassifier(hidden_layer_sizes=(16,8),
    activation = 'relu',
    solver = 'adam',
    learning_rate_init=0.001,
    max_iter=1000,
    random_state=42))
])
#Yekseri setting dadam (badan migam)



````

````python
model.fit(X_train,y_train)

````

````python
y_pred = model.predict(X_test)

from sklearn.metrics import accuracy_score
test_score_acccuracy = accuracy_score(y_test,y_pred)


print(f"Test score accuracy: {test_score_acccuracy}")


````

````python

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.neural_network import MLPClassifier

model = Pipeline([
    ('scaler',StandardScaler()),
    ('mlp',
    MLPClassifier(hidden_layer_sizes=(100,100),
    activation = 'relu',
    solver = 'adam',
    learning_rate_init=0.0001,
    max_iter=10000,
    random_state=42))
])

model.fit(X_train,y_train)

y_pred = model.predict(X_test)

from sklearn.metrics import accuracy_score
test_score_acccuracy = accuracy_score(y_test,y_pred)

print(f"Test score accuracy: {test_score_acccuracy}")

#relu , adam , 0.001 , (16.8) --> 0.95
#tanh , adam , 0.001 , (100,10) --> 0.95
#relu , adam , 0.001 , (100,10) --> 0.96
#relu , adam , 0.001 , (100,100) --> 0.96
````

````python
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    (
        "mlp",
        MLPClassifier(
            max_iter=1000,
            early_stopping=True,
            random_state=42
        )
    )
])

param_grid = {
    "mlp__hidden_layer_sizes": [
        (16,),
        (32,),
        (32, 16),
        (64, 32)
    ],
    "mlp__activation": [
        "relu",
        "tanh"
    ],
    "mlp__solver": [
        "adam",
        "lbfgs"
    ],
    "mlp__alpha": [
        0.0001,
        0.001,
        0.01
    ]
}

grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="accuracy",
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train, y_train)

print("Best parameters:")
print(grid_search.best_params_)

print("Best cross-validation score:")
print(grid_search.best_score_)

print("Test score:")
print(grid_search.score(X_test, y_test))
````

````python
X_train = X_train /255.0
X_test = X_test / 255.0
````

````python
import tensorflow as tf   

model = tf.keras.Sequential([

    tf.keras.layers.Flatten(input_shape=(28,28)),

    tf.keras.layers.Dense(units=128,activation='relu'),

    tf.keras.layers.Dense(units=64,activation='relu'),

    tf.keras.layers.Dense(units=10,activation='softmax')
])

# (28x28) 784 * 128 * 64 * 10 
````

````python
model.summary()
````

````python
model.compile(
    optimizer='adam',
    loss = 'sparse_categorical_crossentropy',
    metrics=['accuracy']
)


#adam,sgd,..--> update wieght
````

train nmikone ha compiler ro set mikone, migi ooptimziation chijori bashe, los chijori , metric

````python
history = model.fit(X_train,y_train, epochs = 10 , batch_size=32, validation_split=0.2)


#10 bar feed + back (epoch) , 32 ta data mifrese too . 20% training data validation 

#10000 test , 60000 training ( 480000 train, 120000 validation)
````

````python
test_loss, test_accuracy = model.evaluate(X_test,y_test)

print('Test Loss:', test_loss)
print('Test Accuracy:',test_accuracy)
````

````python
import torch
import torch.nn as nn
import torch.optim as optim

````

````python
if torch.cuda.is_available():
    device = torch.device('cuda')

elif torch.backends.mps.is_available():
    device=torch.device('mps')
else:
    device = torch.device('cpu')


print('device:',device)
````

````python
from torchvision import datasets, transforms
transform =transforms.ToTensor()


train_datasets = datasets.MNIST(root='./data',train=True, download=True,transform=transform)

test_datasets= datasets.MNIST(root='./data',train=False, download=True,transform=transform)
````

````python
print(type(train_datasets))
````

````python
print(len(train_datasets))
print(len(test_datasets))
````

````python
#ma nemikhahim kole data 60000 baham varede network, bvatch size darim ag yadeton bshe

#batch size tesnorflow compile misashti, inja na ma yechizi darim bename

#data loader (datasets ) behet hey done done data pass mide 

#Pytorch datalaoder

from torch.utils.data import DataLoader

BATCH_SIZE=64

train_loader = DataLoader(train_datasets,batch_size=BATCH_SIZE,shuffle=True)

test_loader = DataLoader(test_datasets,batch_size=BATCH_SIZE,shuffle=True)
````

````python
import torch.nn as nn

class MNISTModelFanavari(nn.Module):

    def __init__(self):
        super().__init__()

        #init --> fght daram chnata variable misazam

        self.flatten = nn.Flatten()

        self.fc1 = nn.Linear(28*28 , 128)  #784 --> 128
        self.relu1 = nn.ReLU()

        self.fc2 = nn.Linear(128,64)
        self.relu2 = nn.ReLU()

        self.fc3= nn.Linear(64,10)


    def forward(self,x):

        #x ro migire
        x = self.flatten(x)

        x = self.fc1(x)
        x = self.relu1(x)

        x= self.fc2(x)
        x = self.relu2(x)

        x= self.fc3(x)

        return x



        
````

````python
model =MNISTModelFanavari()

model = model.to(device)

print(model)
````

````python
criterion = nn.CrossEntropyLoss() #classification  , Regression MSE 
#niazi b softmax 


#sgd, adam , lbfs

#learning rate va optimziier
optimizer =  optim.Adam(model.parameters(), lr=0.001)
````

````python
epochs  = 10



for epoch in range(epochs):

    model.train()

    total_loss=0

    #ltrain loader --> bath 64 64 
    for image,labels in train_loader:

        image = image.to(device)
        labels = labels.to(device)
        # -------------------------
        # 1. Reset gradients
        # -------------------------
        optimizer.zero_grad() 

        # -------------------------
        # 2. Forward
        # -------------------------
        outputs = model(image) #mage -_> forward --> output

        # -------------------------
        # 3. Calculate loss
        # -------------------------
        loss = criterion(outputs, labels)


        # -------------------------
        # 4. Back propagation
        # -------------------------
        loss.backward()

        # -------------------------
        # 5. update weights
        # -------------------------
        optimizer.step()

        total_loss += loss.item()

    average_Loss = total_loss / len(train_loader)

    print(f"Epoch {epoch+1}/{epochs}, Loss: {average_Loss:.4f}")
````

````python
import torch

model.eval()

correct = 0
total = 0

with torch.no_grad():

    for images, labels in test_loader:

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)

        _,predicted = torch.max(outputs,dim=1)

        total += labels.size(0)

        correct += (predicted == labels).sum().item()


accuracy = 100 * correct / total

print(f"Accuracy of the network on the 10000 test images: {accuracy:.2f}%")



````
