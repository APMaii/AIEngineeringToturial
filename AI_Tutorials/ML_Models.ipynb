LinearRegression --> sklearn LinearRegression , SGDRegression

$$a_{new} = a_{old} - \eta * \frac{\delta{loss}}{\delta{a}}$$

## Supervides Machine learning Regressor Models

LinearRegression() ---> miomd a , b --> yek matrix ro hesab mikrd  Y*X-1 ro hesab mikard


SGDRegressor() -> a,b ro random mziad, loss function ro hesab mikrd va baraye minimize krdnesh ba estefade az gradient descent pish miraft.

Loss function = 1/n radicale (prediction - experimental)**2 

loss fucntion mikhad kam kone --> 0 --> prediction ~= experimental


Ridge() Lasso() ElasticNet()




$$Loss function = min \sum{{(y_{i} - \hat{y_{i}})}^{2}}$$


#### Ridge / L2 

$$Loss function = min \sum{{(y_{i} - \hat{y_{i}})}^{2}} + \lambda \sum{w_{i}}^2$$


yani inke bia  vaghty loss fucntion ro hesab mikoni, na tnha donbael in bashi k pishbinit nazdik b vagheiat bashe, terme y - yi ro kam koni, balke bayad inkar ro ba vazn haye koochak tar anjam bdi , chon ag khhob pishbini komni y-yi ~0 ama ba vazn haye bal ainkaro koni (coeffciient ,a,b,c,d,....) loss fucntion y adadd bozorg . inghd talash mikone , Ridge --> penalty (L2) 


kh vaght ha kh az feature (x) ha tasiri nadaran , vaghty az SGDregressor sade estefade mikonim , motasefane ,yekseri vorodi k tasiri nadarsn ham coefficiente (zaribe) balaei migiran , predict badtar mikonan . 


#### Lasso / L1



$$Loss function = min \sum{{(y_{i} - \hat{y_{i}})}^{2}} + \lambda |w_{i}|$$

w haro sefr kone, ridge , bayad zarib haro (a,b,c ) ta hade momken kochak konad va hamchnin pishbinie khobi dashte bashad. lasso (L1) --> sefre sefr mikone, na tanha dare pishbnin mikone, feature selection ham mikone. farda mitoni negah koni . 9 ta feature (x) distance, rain , .... , model age biad dose ta zaribe poshte inaro sefr -> bedard nemikhrodan , feature (vordoi hayte) bi tasir bodan --> feature selection . 


#### Elastic Net (l1 / l2)

MAN yek tarkibi az ridge va lasso hastam , yani ham yeseriaro sefr mikonm ham yeseri haro nazdike 0 negah midadam . va ye darsadi , 80% L1 , 20 % L2 ,....


Penalty -->loss fucntion --> Regularization methods





````python
import numpy as np

# 10 samples, 10 features
# Columns:
# 0: area (m²)
# 1: bedrooms
# 2: bathrooms
# 3: building age (years)
# 4: floor
# 5: parking spots
# 6: distance to city center (km)
# 7: school rating (1-10)
# 8: crime index (lower is better)
# 9: renovation score (0-10)

x_train = np.array([
    [55,  1, 1, 25, 2, 0, 12, 6, 7, 3],   # small old apartment
    [75,  2, 1, 15, 4, 1,  8, 7, 5, 5],   # medium apartment
    [90,  2, 2, 10, 6, 1,  6, 8, 4, 6],   # newer 2BR
    [110, 3, 2,  8, 3, 1,  5, 8, 3, 7],   # family apartment
    [130, 3, 2,  5, 7, 2,  4, 9, 2, 8],   # large renovated unit
    [150, 4, 3, 12, 2, 2,  7, 7, 4, 6],   # big apartment
    [180, 4, 3,  3, 5, 2,  3, 9, 2, 9],   # almost new large unit
    [200, 5, 3,  2, 8, 2,  2, 9, 1, 9],   # premium apartment
    [65,  2, 1, 30, 1, 0, 15, 5, 8, 2],   # old far from center
    [95,  3, 2,  7, 4, 1,  6, 8, 3, 7],   # balanced mid-range
], dtype=float)

# Target: house price (e.g. million Toman)
y_train = np.array([
    3200,   # 55 m², old, far
    4800,   # 75 m²
    6200,   # 90 m²
    7500,   # 110 m²
    9200,   # 130 m² renovated
    8800,   # 150 m²
    11500,  # 180 m² new
    13800,  # 200 m² premium
    2900,   # old + far + high crime
    7100,   # balanced
], dtype=float)
````

````python
from sklearn.linear_model import SGDRegressor

model = SGDRegressor()

model.fit(x_train,y_train)


print(model.coef_) 


#parametr haee k darovorde
#-2 * area -1.39 * bedrooms -1.25*.....
````

````python
from sklearn.linear_model import Ridge

model = Ridge(alpha= 0.1) #harche alpha (landa) bishtar yani hadafe modelet mire vase inke zarib haye koochiktri peyda kone 

model.fit(x_train,y_train)


#alpha --> hyperparameter --> taenesh mikonam l va tasir mizare rooye yadgirie model
#hyperparamter --> fara parameter

model.coef_
````

````python
from sklearn.linear_model import Lasso

model = Lasso(alpha= 0.1) #sefr mikone zarib ha 
#Miad bishrtar zarib haro sefr mikone 


model.fit(x_train,y_train)

model.coef_

#inja sefr ndrim , mire donbale inke bishtr pishbinie behtri kone , 
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns



x = np.arange(0,200,20)
y = 4 * x + np.random.randn(10) * 73


from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x,y, test_size=0.2, random_state=42)





plt.scatter(x_train,y_train,label='train')
plt.scatter(x_test,y_test,label='test')
plt.ylabel('y')
plt.xlabel('x')
plt.legend()
plt.show()











````

````python
from sklearn.linear_model import LinearRegression


model = LinearRegression()

model.fit(x_train.reshape(-1,1),y_train)


y_train_pred = model.predict(x_train.reshape(-1,1))


y_test_pred = model.predict(x_test.reshape(-1,1))




plt.scatter(x_train,y_train,label='true train')
plt.plot(x_train,y_train_pred,label='predict train')
plt.ylabel('y')
plt.xlabel('x')
plt.legend()
plt.show()

````

yekseri noghat daram k esmeshon has train dataset (x_train, y_train ) hamchnin yekseri khat daram --> modele mane k train dide 

model.predict(x_train) ---> x_train_pred


````python

plt.scatter(x_train,y_train,label='true train')
plt.scatter(x_train,y_train_pred,label='predict train')
plt.ylabel('y')
plt.xlabel('x')
plt.legend()
plt.show()
````

````python
from sklearn.metrics import mean_absolute_error


mae_train_score = mean_absolute_error(y_train, y_train_pred)

print(f"MAE train score: {mae_train_score}")
````
